# Step 5：Half Skew与Wing Curvature期限结构

本模块只消费04生成的 `half_skew_metrics`，绘制并保存：

- \(CallSkew=\sigma(h)-\sigma(0)\)
- \(PutSkew=\sigma(-h)-\sigma(0)\)
- \(WingCurvature=(CallSkew+PutSkew)/2\)

不在05重新拟合波动率。

```python
from pathlib import Path
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
from IPython.display import Image,display
plt.style.use('seaborn-v0_8-whitegrid')
```

## 数据整理

```python
def prepare_half_skew_term_structure(source):
    required=['TRADE_DT','EXPIRY','EXPIRY_CODE','HALF_SKEW_H','CALL_SKEW','PUT_SKEW','WING_CURVATURE']
    missing=[c for c in required if c not in source.columns]
    if missing: raise KeyError(f'half_skew_metrics缺少字段: {missing}')
    result=source[required].copy()
    result.TRADE_DT=pd.to_datetime(result.TRADE_DT,errors='coerce').dt.normalize()
    result.EXPIRY=pd.to_datetime(result.EXPIRY,errors='coerce').dt.normalize()
    numeric=['HALF_SKEW_H','CALL_SKEW','PUT_SKEW','WING_CURVATURE']
    result[numeric]=result[numeric].apply(pd.to_numeric,errors='coerce')
    invalid=result[required].isna().any(axis=1)|~np.isfinite(result[numeric]).all(axis=1)
    if invalid.any(): warnings.warn(f'删除{int(invalid.sum())}行无效记录'); result=result.loc[~invalid].copy()
    if result.duplicated(['TRADE_DT','EXPIRY']).any(): raise ValueError('存在重复TRADE_DT与EXPIRY')
    if not np.allclose(result.WING_CURVATURE,.5*(result.CALL_SKEW+result.PUT_SKEW)):
        raise ValueError('Wing Curvature不满足定义')
    return result.sort_values(['TRADE_DT','EXPIRY']).reset_index(drop=True)

half_skew_term_structure=prepare_half_skew_term_structure(half_skew_metrics)
display(half_skew_term_structure.head())
print(f'Rows: {len(half_skew_term_structure)}, Expiries: {half_skew_term_structure.EXPIRY.nunique()}')
```

## 绘图与保存

```python
COLORS={'CALL_SKEW':'#2b6cb0','PUT_SKEW':'#d1495b','WING_CURVATURE':'#6b46c1'}
LABELS={'CALL_SKEW':'Call Skew','PUT_SKEW':'Put Skew','WING_CURVATURE':'Wing Curvature'}

def plot_metric_all_expiries(data,metric,output_file):
    output_file=Path(output_file); output_file.parent.mkdir(parents=True,exist_ok=True)
    fig,ax=plt.subplots(figsize=(13,6.5))
    for expiry in sorted(data.EXPIRY.unique()):
        sample=data.loc[data.EXPIRY.eq(expiry)].sort_values('TRADE_DT')
        ax.plot(sample.TRADE_DT,sample[metric],marker='o',ms=3,lw=1.5,label=pd.Timestamp(expiry).strftime('%Y%m'))
    ax.axhline(0,color='#697386',ls='--',lw=.9)
    ax.set_title(f'{LABELS[metric]} - All Expiries | h={float(data.HALF_SKEW_H.iloc[0]):.3f}',fontsize=14,fontweight='bold')
    ax.set_xlabel('Trade Date'); ax.set_ylabel(LABELS[metric]); ax.yaxis.set_major_formatter(PercentFormatter(1.0))
    ax.legend(title='Expiry',ncol=min(4,data.EXPIRY.nunique())); ax.grid(color='#d9e0e8',lw=.7,alpha=.8)
    fig.autofmt_xdate(); fig.tight_layout(); fig.savefig(output_file,dpi=170,bbox_inches='tight',facecolor='white'); plt.close(fig)
    return output_file

def plot_half_skew_by_expiry(data,output_dir,figure_format='png'):
    output_dir=Path(output_dir); output_dir.mkdir(parents=True,exist_ok=True); saved={}
    for expiry in sorted(data.EXPIRY.unique()):
        sample=data.loc[data.EXPIRY.eq(expiry)].sort_values('TRADE_DT'); code=pd.Timestamp(expiry).strftime('%Y%m')
        fig,ax=plt.subplots(figsize=(11,5.5))
        for metric in ['CALL_SKEW','PUT_SKEW']:
            ax.plot(sample.TRADE_DT,sample[metric],color=COLORS[metric],marker='o',ms=3.5,lw=1.6,label=LABELS[metric])
        ax.axhline(0,color='#697386',ls='--',lw=.9)
        ax.set_title(f'Call and Put Half Skew - {code} | h={float(sample.HALF_SKEW_H.iloc[0]):.3f}',fontsize=13,fontweight='bold')
        ax.set_xlabel('Trade Date'); ax.set_ylabel('Half skew (IV difference vs ATM)'); ax.yaxis.set_major_formatter(PercentFormatter(1.0))
        ax.legend(); ax.grid(color='#d9e0e8',lw=.7,alpha=.8); fig.autofmt_xdate(); fig.tight_layout()
        path=output_dir/f'half_skew_{code}.{figure_format}'; fig.savefig(path,dpi=170,bbox_inches='tight',facecolor='white'); plt.close(fig); saved[code]=path
    return saved

def save_half_skew_term_structure(data,output_file):
    output_file=Path(output_file); output_file.parent.mkdir(parents=True,exist_ok=True)
    data.to_csv(output_file,index=False,encoding='utf-8-sig',date_format='%Y-%m-%d'); return output_file
```

## 执行与结果

```python
output_path=Path(SKEW_CURVATURE_OUTPUT_PATH); figure_path=output_path/'figures'; data_path=output_path/'data'
all_expiry_files={}
if SAVE_FIGURE:
    for metric in ['CALL_SKEW','PUT_SKEW','WING_CURVATURE']:
        all_expiry_files[metric]=plot_metric_all_expiries(
            half_skew_term_structure,metric,figure_path/f'{metric.lower()}_all_expiries.{FIGURE_FORMAT}')
half_skew_by_expiry_files=plot_half_skew_by_expiry(
    half_skew_term_structure,figure_path/'half_skew_by_expiry',FIGURE_FORMAT) if SAVE_FIGURE else {}
half_skew_term_structure_file=save_half_skew_term_structure(
    half_skew_term_structure,data_path/'half_skew_term_structure.csv') if SAVE_CSV else None
print(f'Output: {output_path}')
print(f'All-expiry figures: {all_expiry_files}')
print(f'By-expiry figures: {len(half_skew_by_expiry_files)}')
print(f'Data file: {half_skew_term_structure_file}')
if all_expiry_files: display(Image(filename=str(all_expiry_files['CALL_SKEW'])))
if half_skew_by_expiry_files:
    example=sorted(half_skew_by_expiry_files)[0]; print(f'Example expiry: {example}'); display(Image(filename=str(half_skew_by_expiry_files[example])))
```
